2.1.Làm sạch dữ liệu

```python
# =====================================================================
# BƯỚC 2: KHẢO SÁT DỮ LIỆU (Data Profiling)
# =====================================================================
# Tạo các bản sao để thực hiện biến đổi dữ liệu sạch
df_time = time_series.copy()
df_demo = demographics.copy()
df_asylum = asylum_seekers.copy()
df_resettle = resettlement.copy()
df_poc = persons_of_concern.copy()

def survey_dataset(name, df):
    print(f"=== KHẢO SÁT TẬP DỮ LIỆU: {name.upper()} ===")
    print(f"- Kích thước: {df.shape[0]} dòng, {df.shape[1]} cột")
    print("- Kiểu dữ liệu cột:")
    print(df.dtypes.to_dict())
    missing_pct = (df.isnull().sum() / len(df)) * 100
    if missing_pct.sum() > 0:
        print("- Tỷ lệ thiếu dữ liệu ban đầu (%):")
        print(missing_pct[missing_pct > 0].to_dict())
    else:
        print("- Không có dữ liệu trống.")
    print("\n")

survey_dataset('time_series', df_time)
survey_dataset('demographics', df_demo)
```

```python
# =====================================================================
# BƯỚC 3: CHUẨN HÓA KIỂU DỮ LIỆU (Data Type Standardization)
# =====================================================================
# Các ký tự ẩn chứa giá trị rỗng như '*', '-', 'unknown' được đưa về chuẩn NaN của numpy để đồng bộ hóa việc xử lý.
def standardize_to_nan(df):
    df_standard = df.copy()
    for col in df_standard.columns:
        if df_standard[col].dtype == 'object':
            df_standard[col] = df_standard[col].astype(str).str.strip()
            df_standard[col] = df_standard[col].replace({'*': np.nan, '-': np.nan, 'nan': np.nan, 'Unknown': np.nan, '': np.nan})
    return df_standard

print("[BƯỚC 3] Đang chuẩn hóa định dạng chuỗi và đưa các ký tự rỗng ẩn về dạng NaN...")
df_time = standardize_to_nan(df_time)
df_demo = standardize_to_nan(df_demo)
df_asylum = standardize_to_nan(df_asylum)
df_resettle = standardize_to_nan(df_resettle)
df_poc = standardize_to_nan(df_poc)
print("Chuẩn hóa hoàn tất.")
```

```python
# =====================================================================
# BƯỚC 4: XỬ LÝ GIÁ TRỊ THIẾU (Handling Missing Values)
# =====================================================================
# Chuyển các cột giá trị thống kê số lượng sang dạng float/int, đồng thời điền khuyết thiếu bằng 0.
# Điền các giá trị phân loại bị thiếu bằng nhãn 'Unknown' để bảo toàn phân loại dòng.

def fill_missing_values(df, numeric_cols, categorical_cols):
    df_filled = df.copy()
    for col in numeric_cols:
        if col in df_filled.columns:
            df_filled[col] = pd.to_numeric(df_filled[col], errors='coerce').fillna(0).astype(int)
    for col in categorical_cols:
        if col in df_filled.columns:
            df_filled[col] = df_filled[col].fillna('Unknown')
    return df_filled

print("[BƯỚC 4] Thực hiện xử lý giá trị thiếu cho các bảng dữ liệu...")
df_time = fill_missing_values(df_time, ['Value'], ['Country / territory of asylum/residence', 'Origin', 'Population type'])

demo_num = ['Female 0-4', 'Female 5-11', 'Female 12-17', 'Female 18-59', 'Female 60+', 'Female total', 'Male 0-4', 'Male 5-11', 'Male 12-17', 'Male 18-59', 'Male 60+', 'Male total', 'Total']
df_demo = fill_missing_values(df_demo, demo_num, ['Country / territory of asylum/residence', 'Location Name'])

asylum_num = ['Applied during year', 'decisions_recognized', 'Rejected', 'Total pending end-year']
df_asylum = fill_missing_values(df_asylum, asylum_num, ['Country / territory of asylum/residence', 'Origin'])

df_resettle = fill_missing_values(df_resettle, ['Value'], ['Country / territory of asylum/residence', 'Origin'])

poc_num = ['Refugees (incl. refugee-like situations)', 'Asylum-seekers', 'Returned refugees', 'IDPs', 'Stateless', 'Others of concern', 'Total population']
df_poc = fill_missing_values(df_poc, poc_num, ['Country / territory of asylum/residence', 'Origin'])
print("Hoàn thành xử lý điền khuyết dữ liệu.")
```

```python
# =====================================================================
# BƯỚC 5: KHỬ DỮ LIỆU TRÙNG LẶP (Deduplication)
# =====================================================================
def drop_duplicates_dataset(name, df):
    initial = len(df)
    df_cleaned = df.drop_duplicates()
    print(f"- {name}: Loại bỏ {initial - len(df_cleaned)} dòng trùng lặp.")
    return df_cleaned

print("[BƯỚC 5] Tiến hành rà soát và xóa các dòng dữ liệu bị trùng...")
df_time = drop_duplicates_dataset('time_series', df_time)
df_demo = drop_duplicates_dataset('demographics', df_demo)
df_asylum = drop_duplicates_dataset('asylum_seekers', df_asylum)
df_resettle = drop_duplicates_dataset('resettlement', df_resettle)
df_poc = drop_duplicates_dataset('persons_of_concern', df_poc)
```

```python
# =====================================================================
# BƯỚC 6: XỬ LÝ GIÁ TRỊ KHÔNG HỢP LỆ (Outliers & Anomalies)
# =====================================================================
# - Đưa các giá trị thống kê số lượng âm (nếu có do sai lệch nhập liệu) về 0.
# - Thiết lập năm báo cáo hợp lệ nằm trong khoảng từ năm 1950 tới hiện tại (2026).

def clean_anomalies_and_years(df, numeric_cols, year_col='Year'):
    df_valid = df.copy()
    for col in numeric_cols:
        if col in df_valid.columns:
            df_valid.loc[df_valid[col] < 0, col] = 0

    if year_col in df_valid.columns:
        df_valid = df_valid[(df_valid[year_col] >= 1950) & (df_valid[year_col] <= 2026)]
    return df_valid

print("[BƯỚC 6] Khử các giá trị âm và chuẩn hóa giới hạn năm báo cáo...")
df_time = clean_anomalies_and_years(df_time, ['Value'])
df_demo = clean_anomalies_and_years(df_demo, demo_num)
df_asylum = clean_anomalies_and_years(df_asylum, asylum_num)
df_resettle = clean_anomalies_and_years(df_resettle, ['Value'])
df_poc = clean_anomalies_and_years(df_poc, poc_num)
print("Hoàn tất xử lý ngoại lai.")
```

```python
# =====================================================================
# BƯỚC 7: KIỂM ĐỊNH TỰ ĐỘNG (Automated Validation)
# =====================================================================
print("[BƯỚC 7] Tiến hành kiểm định chất lượng dữ liệu tự động...")

try:
    # 1. Kiểm tra khuyết thiếu cho cột chính 'Value' trong bảng time_series
    assert df_time['Value'].isnull().sum() == 0, "Lỗi: Cột 'Value' vẫn còn chứa giá trị rỗng!"

    # 2. Kiểm tra tính hợp lệ về dấu cho 'Value'
    assert (df_time['Value'] >= 0).all(), "Lỗi: Phát hiện giá trị âm bất thường trong bảng 'time_series'!"

    # 3. Kiểm tra tính trùng lặp
    assert df_time.duplicated().sum() == 0, "Lỗi: Vẫn còn dòng trùng lặp chưa được loại bỏ sạch!"

    print("\n>>> KIỂM ĐỊNH TỰ ĐỘNG THÀNH CÔNG 100%! <<<\nDữ liệu đã đạt tiêu chuẩn chất lượng cao để tiếp tục phân tích trực quan hóa.")
except AssertionError as error_msg:
    print(f"\n❌ KIỂM ĐỊNH KHÔNG THÔNG QUA: {error_msg}")
```

### 3.2. Trực quan hóa luồng di chuyển (Flow Analysis)

Phân tích và biểu diễn luồng di chuyển trực quan từ quốc gia/châu lục nguồn đến quốc gia/châu lục tiếp nhận.

**Biểu đồ sử dụng:** Biểu đồ dòng chảy Sankey Diagram thể hiện quy mô và hướng di chuyển giữa các khu vực; Biểu đồ cột nhóm/cột chồng (Grouped/Stacked Barplot) theo châu lục.



### 3.3. Trực quan hóa phân phối & Mật độ di cư

Phân tích sự phân bố quy mô di cư giữa các quốc gia (kiểm tra độ lệch của phân phối quy mô tị nạn).

**Biểu đồ sử dụng:** Biểu đồ tần suất (Histogram), Biểu đồ hộp (Boxplot), Biểu đồ mật độ (Density Plot).



### 3.4. Trực quan hóa mối liên hệ & Ma trận tương quan

Xây dựng ma trận nhiệt (Heatmap) thể hiện mức độ di cư giữa các khu vực địa lý chính trên thế giới.



## 4. Phân tích và thống kê



### 4.1. Phân tích và thống kê xu hướng di cư theo thời gian

**Mục đích:** Đánh giá sự biến động tổng lượng người tị nạn/di cư qua các năm để phát hiện các giai đoạn bùng nổ (peaks) gắn liền với các sự kiện địa - chính trị hoặc xung đột lịch sử trên thế giới.

**Cách thực hiện:**
- Sử dụng hàm gom nhóm `groupby('Year')` để tính tổng số lượng người tị nạn, đơn đăng ký tị nạn và số lượng hồ sơ được chấp nhận/tồn đọng qua từng năm.
- Vẽ biểu đồ đường (Line Plot) biểu diễn chuỗi thời gian xu hướng di cư toàn cầu.
- Tính toán tỷ lệ tăng trưởng hàng năm (YoY Growth Rate) và chỉ số biến động nhân khẩu học theo thời gian.



### 4.2. Phân tích cơ cấu nhân khẩu học & kết quả xử lý đơn tị nạn

**Mục đích:** Nhận diện độ tuổi, giới tính yếu thế (trẻ em, phụ nữ) và đánh giá tỷ lệ duyệt/từ chối hồ sơ tị nạn giữa các quốc gia.

**Cách thực hiện:**
- Trực quan hóa tỷ lệ giới tính (Nam/Nữ) và nhóm tuổi (0–4, 18–59, 60+) bằng biểu đồ Pie chart hoặc Grouped Bar chart.
- Trích xuất tỷ lệ hồ sơ được chấp nhận (decisions_recognized) so với bị từ chối (Rejected) và còn tồn đọng (Total pending end-year) để phân tích hiệu quả xử lý đơn của từng quốc gia tiếp nhận.



## 5. Chủ đề nâng cao



### Chủ đề nâng cao: Phân cụm dữ liệu (K-Means Clustering)

**Mục đích:** Phân nhóm tự động các quốc gia dựa trên đặc tính tương đồng về gánh nặng tiếp nhận hoặc quy mô xuất khẩu người tị nạn, hỗ trợ phân bổ nguồn lực cứu trợ.

**Các bước thực hiện:**
1. **Trích xuất đặc trưng (Feature Engineering):** Tổng hợp dữ liệu theo quốc gia với các chỉ số: Tổng số lượng tiếp nhận, Tổng số lượng xuất khẩu, Tỷ lệ chấp nhận đơn, Tổng số lượng tồn đọng, và tỷ lệ phụ nữ/trẻ em.
2. **Chuẩn hóa dữ liệu:** Sử dụng StandardScaler hoặc MinMaxScaler để đưa các biến định lượng về cùng một thang đo.
3. **Xác định số cụm tối ưu:** Áp dụng phương pháp khuỷu tay (Elbow Method) và chỉ số Silhouette Score để tìm số lượng cụm phù hợp.
4. **Huấn luyện mô hình:** Chạy thuật toán K-Means Clustering và gán nhãn phân cụm cho từng quốc gia.
5. **Trực quan & Đánh giá:** Trực quan hóa các cụm bằng biểu đồ phân tán (Scatter Plot) sau khi giảm chiều dữ liệu (PCA/t-SNE) và đưa ra nhận xét về đặc điểm từng cụm quốc gia.
