### 2.1. Làm sạch và Chuẩn hóa dữ liệu
Dưới đây là mã nguồn để xử lý các giá trị khuyết, chuẩn hóa kiểu dữ liệu số, và loại bỏ các dữ liệu trùng lặp trong các bảng.

```python
import numpy as np
import pandas as pd

def clean_and_convert_numeric(df, numeric_cols):
    """
    Hàm thay thế các ký tự đặc biệt (*, -) thành NaN,
    sau đó chuyển cột sang dạng số (float/int) và điền khuyết bằng 0.
    """
    df_cleaned = df.copy()
    for col in numeric_cols:
        if col in df_cleaned.columns:
            # Thay thế ký tự đặc biệt bằng NaN
            df_cleaned[col] = df_cleaned[col].astype(str).str.strip()
            df_cleaned[col] = df_cleaned[col].replace({'*': np.nan, '-': np.nan, 'nan': np.nan, '': np.nan})
            # Chuyển đổi sang kiểu số
            df_cleaned[col] = pd.to_numeric(df_cleaned[col], errors='coerce')
            # Điền giá trị thiếu bằng 0 cho các cột thống kê số lượng
            df_cleaned[col] = df_cleaned[col].fillna(0).astype(int)
    return df_cleaned

# --- 1. Làm sạch bảng time_series ---
print("Đang làm sạch 'time_series'...")
time_series_clean = clean_and_convert_numeric(time_series, ['Value'])
time_series_clean = time_series_clean.drop_duplicates()

# --- 2. Làm sạch bảng demographics ---
print("Đang làm sạch 'demographics'...")
demo_numeric_cols = [
    'Female 0-4', 'Female 5-11', 'Female 12-17', 'Female 18-59', 'Female 60+', 'Female total',
    'Male 0-4', 'Male 5-11', 'Male 12-17', 'Male 18-59', 'Male 60+', 'Male total', 'Total'
]
demographics_clean = clean_and_convert_numeric(demographics, demo_numeric_cols)
demographics_clean = demographics_clean.drop_duplicates()

# --- 3. Làm sạch bảng asylum_seekers ---
print("Đang làm sạch 'asylum_seekers'...")
asylum_numeric_cols = ['Applied during year', 'decisions_recognized', 'Rejected', 'Total pending end-year']
asylum_seekers_clean = clean_and_convert_numeric(asylum_seekers, asylum_numeric_cols)
asylum_seekers_clean = asylum_seekers_clean.drop_duplicates()

# --- 4. Làm sạch bảng resettlement ---
print("Đang làm sạch 'resettlement'...")
resettlement_clean = clean_and_convert_numeric(resettlement, ['Value'])
resettlement_clean = resettlement_clean.drop_duplicates()

# --- 5. Làm sạch bảng persons_of_concern ---
print("Đang làm sạch 'persons_of_concern'...")
poc_numeric_cols = [
    'Refugees (incl. refugee-like situations)', 'Asylum-seekers',
    'Returned refugees', 'IDPs', 'Stateless', 'Others of concern', 'Total population'
]
persons_of_concern_clean = clean_and_convert_numeric(persons_of_concern, poc_numeric_cols)
persons_of_concern_clean = persons_of_concern_clean.drop_duplicates()

print("\n--- HOÀN THÀNH LÀM SẠCH DỮ LIỆU ---")
```

## 5. Chủ đề nâng cao



### Chủ đề nâng cao: Phân cụm dữ liệu (K-Means Clustering)

**Mục đích:** Phân nhóm tự động các quốc gia dựa trên đặc tính tương đồng về gánh nặng tiếp nhận hoặc quy mô xuất khẩu người tị nạn, hỗ trợ phân bổ nguồn lực cứu trợ.

**Các bước thực hiện:**
1. **Trích xuất đặc trưng (Feature Engineering):** Tổng hợp dữ liệu theo quốc gia với các chỉ số: Tổng số lượng tiếp nhận, Tổng số lượng xuất khẩu, Tỷ lệ chấp nhận đơn, Tổng số lượng tồn đọng, và tỷ lệ phụ nữ/trẻ em.
2. **Chuẩn hóa dữ liệu:** Sử dụng StandardScaler hoặc MinMaxScaler để đưa các biến định lượng về cùng một thang đo.
3. **Xác định số cụm tối ưu:** Áp dụng phương pháp khuỷu tay (Elbow Method) và chỉ số Silhouette Score để tìm số lượng cụm phù hợp.
4. **Huấn luyện mô hình:** Chạy thuật toán K-Means Clustering và gán nhãn phân cụm cho từng quốc gia.
5. **Trực quan & Đánh giá:** Trực quan hóa các cụm bằng biểu đồ phân tán (Scatter Plot) sau khi giảm chiều dữ liệu (PCA/t-SNE) và đưa ra nhận xét về đặc điểm từng cụm quốc gia.
